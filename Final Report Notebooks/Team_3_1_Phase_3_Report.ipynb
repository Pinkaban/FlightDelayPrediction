## Abstract

This project tackles the high-impact problem of predicting flight departure delays of 15 minutes or more—a challenge that costs the U.S. airline industry over $30 billion annually and affects millions of passengers. While earlier phases established a baseline using the pre-joined OTPW dataset and Logistic Regression (best F2=0.5500), Phase III introduces a major step forward through both data engineering and model development.

The central innovation is the construction of a custom flight–weather dataset, replacing the static OTPW join with a more precise integration of BTS flight records and minute-level NOAA weather observations. This custom join leverages station-level geographic matching and a 2-hour pre-departure window, enriched with engineered features capturing temporal patterns, weather dynamics, operational behavior, network delay propagation, and aircraft turnaround effects. From an initial pool of over 200 candidate variables, a systematic feature selection process identified the top 20 most predictive features using L2 regularization method, balancing model performance, interpretability, and robustness.

Using this enhanced dataset (spanning over 29 million flights from 2015–2019), we evaluated multiple modeling approaches, including Logistic Regression, Multilayer Perceptron (MLP), LSTM, and XGBoost, under a consistent framework with temporal validation, class imbalance handling, and F2-optimized threshold tuning.

Results show a substantial performance improvement across all models when using the custom-joined data. While XGBoost emerged as the strongest individual model (F2=0.826 on the 2019 blind test set), the best overall performance was achieved through an ensemble (stacking) approach combining Logistic Regression and XGBoost. This final model improved the F2 score to 0.8285, with a precision of 0.6185 and a recall of 0.9054, further enhancing the model’s ability to correctly identify delayed flights. These results underscore the value of combining complementary modeling approaches, as well as the critical role of high-quality data integration and targeted feature selection in unlocking predictive performance beyond what model complexity alone can achieve.

Overall, this phase demonstrates that combining domain-informed feature engineering with modern machine learning techniques can significantly enhance delay prediction accuracy, providing actionable insights for airlines to mitigate operational disruptions.

### Project Description
Flight delays driven by operational bottlenecks, technical aircraft issues, and adverse weather represent a massive logistical challenge. The aviation industry faces over $30 billion in annual costs from flight delays in the U.S. alone, with the global economic impact of disruptions reaching up to $67.5 billion when factoring in lost passenger time and operational overhead. These expenses include wasted fuel, crew time, maintenance, and lost productivity (Anupkumar, 2023).

Our analysis draws on multiple complementary datasets spanning 2015–2019. In Phase III, we migrated from the pre-joined OTPW (On Time Performance and Weather) dataset used in Phase II to a custom flight–weather integrated dataset. This custom join was built by directly integrating U.S. DOT Bureau of Transportation Statistics (BTS) flight records with NOAA hourly weather observations, FAA station metadata, and airport reference data. The custom join provides several advantages over OTPW: (1) minute-level time alignment of weather to flights using a 2-hour pre-departure window, (2) neighbor-station weather fallback for improved feature completeness, (3) sliding-window weather accumulations capturing evolving conditions, (4) airport network delay propagation features, and (5) multi-year coverage enabling proper temporal cross-validation with 2019 as a blind test set.

The target variable remains DEP_DEL15, a binary indicator that equals 1 when a flight departure delay exceeds 15 minutes. The dataset exhibits a natural class imbalance of approximately 80 percent on-time flights to 20 percent delayed flights. Through expanded feature engineering, we grew the feature space to approximately 100 model-ready features organized into five families: temporal features (quarter, month, day of week, departure hour, holiday proximity), meteorological features (wind speed, visibility, precipitation, accumulated weather windows, derived weather indicators), operational features (departure sequencing, cumulative delay rates, multi-horizon delay propagation), graph-based features (airline-level PageRank capturing network centrality and connectivity), and time-series features (recency, frequency, and monetary-style delay aggregations capturing recent delay behavior).

We frame this as a binary classification problem, predicting whether a given flight will be delayed by 15 minutes or more at departure. Our methodology employs three modeling pipelines: (1) distributed Logistic Regression with L2 regularization as the continued baseline, (2) XGBoost for capturing nonlinear interactions and feature importance, and (3) a Multilayer Perceptron (MLP) Neural Network and (4) LSTM for learning complex decision boundaries. All models use strict chronological temporal splits with 2015–2018 for training/validation and 2019 as the blind test set, fixed random undersampling on the full training set prior to cross-validation folding, sliding-window time-series cross-validation, and F2-optimized threshold tuning.

Our data pipeline is built from four primary, authoritative sources, each contributing complementary signals to support robust modeling.

* **Flight Operations Data** from the U.S. Department of Transportation forms the backbone of the dataset, containing ~63M records for years 2015-2019 with 108 features. This transactional layer includes unique flight identifiers, schedules, carriers, origins/destinations, and operational outcomes.
* **Weather Data** sourced from National Oceanic and Atmospheric Administration provides ~639M records for years 2015-2019 with 123 features, capturing detailed atmospheric conditions at origin and destination locations. These variables add critical environmental context that strongly influences flight performance.
* **Airport Reference Data** (IATA codes and geospatial coordinates) contributes ~57K records with 12 features, enabling standardized location mapping and spatial joins across datasets.
* **Station Network Data** includes ~5M records with 12 features, representing relationships among 2,237 weather stations and their neighbors, supporting spatial interpolation and localized weather feature enrichment.

### Custom Join Pipeline

To produce a richer, more temporally precise feature set, we implemented a cusom join pipeline integrating the source datasets. The custom join pipeline orchestrates the end-to-end integration of flight, weather, station, and airport data into a unified modeling dataset. It begins with systematic data preparation, deduplicating flight records and cleaning stations, weather, and airport datasets followed by feature construction for both weather conditions and flight delay signals. These features are then spatially and temporally aligned by linking each flight to its nearest origin and destination weather stations and incorporating weather observations from a defined pre-departure window (two hours prior). The pipeline further enriches the dataset with graph-based features that capture network-level delay propagation across airports. The final output is a consolidated, feature-rich dataset designed to support robust delay prediction modeling.


![Custom Join Pipeline Diagram](https://raw.githubusercontent.com/Pinkaban/FlightDelayPrediction/refs/heads/main/Final%20Report%20Notebooks/ReportImages/Phase%20III%20Report%20Images/w261_project_phase3_sthakur_custom_data_integration.png)

#### Key Data Transformations

The custom join pipeline designed to address limitations of the OTPW dataset includes following key steps:

**Step 1: Prepare Input Datasets**
* Filtered weather to US stations, removed high-missingness columns (>94% null), deduplicated records
* Filtered airports to US, open, with valid IATA codes (1,972 airports)
* Built station-neighbor lookup (distance ≤ 80 km) for weather fallback

**Step 2: Build Weather Features for Each Station**
* Base features with neighbor-station fallback (metric-level, not station-level)
* Sliding time-window weather accumulation (3h, 6h, 12h, daily) for precipitation, wind, temperature, visibility, humidity
* Daily weather aggregation from midnight to 2 hours before departure

**Step 3: Align Weather to Flights**
* Each flight mapped to weather with max timestamp before the cutoff (2 hours before scheduled departure)
* This ensures no future weather data leaks into predictions

**Step 4: Add Flight-Level Operational Features**
* Airport-level delay features with max timestamp before the cutoff (1h, 3h, 6h, daily)
* Tail-number turnaround metrics
* Route-level delay history

**Step 5: Add Graph and Time-Series Features**
* Airline-level PageRank from the airport connectivity graph
* RFM-style delay recency, frequency, and monetary aggregations
* Holiday proximity and event-based indicators

## Data and Feature Engineering

The predictive performance of flight delay models depends heavily on the quality and contextual richness of engineered features. To capture the multifactorial nature of departure delays, features were constructed by integrating four primary datasets: flight operations, weather observations, station mappings, and airport network metadata using the custom jooin pipeline.

Features were engineered to capture both immediate operational state and historical temporal dependencies, enabling the models to learn short-term disruption signals as well as long-range propagation effects.

### Summary of EDA and Feature Engineering

The extensive EDA conducted across Flights, Weather, Airport Codes, Stations, and OTPW datasets revealed several key findings that informed our feature engineering:

* **Flight delays are primarily carrier-driven:** Carrier-related issues account for nearly half of all delays, while weather contributes about 3%

* **Airline- and airport-specific effects:** Frontier Airlines (F9) has the highest delay rate (∼51% above average); hub airports like ORD reach 30.4% delay rates
* **Domino-effect variables:** Intra-day and inter-day delay propagation features are strong predictors
* **Weather impacts:** Extreme conditions (high winds, low visibility, temperature extremes) increase delay likelihood
* **Data quality:** Careful deduplication and outlier treatment were essential for clean joins

### Feature Family 1: Temporal Features
Temporal features were engineered to capture the time-dependent structure of flight departure delays, reflecting how operational performance varies across multiple time horizons including intraday schedules, weekly cycles, monthly seasonality, annual effects, and holiday-driven demand surges. These features provide the foundational context for understanding when delays are most likely to occur, independent of weather or network effects. A key design objective was to distinguish between coarse seasonal effects (e.g., quarter-level trends) and finer-grained temporal signals (e.g., monthly and weekday variations), while also quantifying nonlinear proximity effects around holidays, which often induce sharp but short-lived operational disruptions.


| Feature Name              | Description                                               | Type                                | Granularity / Scale    | Business Relevance                                                               |
| ------------------------- | --------------------------------------------------------- | ----------------------------------- | ---------------------- | -------------------------------------------------------------------------------- |
| `QUARTER`                 | Calendar quarter in which the flight is scheduled (Q1–Q4) | Categorical (ordinal)               | Seasonal / quarterly   | Captures broad seasonal traffic and weather trends                               |
| `month`                   | Month of scheduled departure (1–12)                       | Categorical (ordinal)               | Monthly                | Models monthly seasonality such as winter storms and summer congestion           |
| `DAY_OF_MONTH`            | Day number within the month (1–31)                        | Numeric / ordinal                   | Daily                  | Captures monthly cyclic operational patterns and holiday adjacency               |
| `DAY_OF_WEEK`             | Day of the week (1–7)                                     | Categorical (ordinal)               | Weekly                 | Captures weekday vs weekend demand and business travel patterns                  |
| `CRS_DEP_TIME`            | Scheduled departure time in HHMM format                   | Numeric (continuous / time-encoded) | Intraday / hourly      | One of the strongest temporal predictors; captures peak-hour congestion risk     |
| `year`                    | Calendar year of flight                                   | Numeric / ordinal                   | Annual                 | Captures long-term trends, structural shifts, and post-event operational changes |
| `is_us_holiday`           | Binary flag indicating US federal holiday                 | Binary categorical                  | Event-based            | Captures holiday surge traffic and irregular scheduling behavior                 |
| `days_to_prev_holiday`    | Number of days since previous holiday                     | Numeric                             | Relative time distance | Measures residual post-holiday congestion effects                                |

#### Key Insights
* The exploratory analysis of temporal features reveals clear and interpretable delay patterns driven by seasonality, weekly travel behavior, and holiday proximity effects.

* In particular, initial exploratory analysis indicates that:

  - Delay behavior exhibits clear seasonality at the monthly level, with pronounced peaks in summer months.
  - There are strong weekly operational cycles, with systematic differences between business days and weekends.
  - Holiday periods introduce localized but high-intensity disruption windows, followed by rapid normalization.
  - Finer-grained calendar features such as day-of-month contribute relatively limited predictive signal.

* An important modeling insight from this analysis is that the temporal effects are clearly nonlinear rather than strictly linear. The monthly, weekly, and holiday curves all show distinct peaks and troughs rather than smooth directional relationships. This indicates that these features should be encoded appropriately (e.g., one-hot encoding, cyclical encoding, or tree-based splits) so that the models can capture their nonlinear structure effectively.

* Overall, temporal features establish the baseline behavioral structure of flight delays, serving as a critical first-order signal layer upon which operational, meteorological, and network-based effects are later superimposed.

### Feature Family 2: Operational Features

Operational features capture the internal dynamics of the aviation system, modeling how delays emerge and propagate due to airport congestion, airline-specific behavior, route-level pressure, aircraft turnaround effects, and short-term network interactions. Unlike temporal features, which describe when flights occur, operational features explain how the system is behaving at that time.

This feature group represents the core predictive layer of the dataset, as it directly reflects real-time and lagged system states that drive delay formation. These signals are primarily derived from flight history, aggregated airport performance, route-level statistics, and propagation effects across aircraft and neighboring airports.

A key design principle in this category was to capture both:

* Immediate operational load (e.g., recent delays, congestion, flight density)
* Delay propagation effects (how delays accumulate and spread across time, routes, and aircraft tails)
* Network-local interactions (neighboring airport influence and shared congestion effects)

These features operate at multiple temporal resolutions (hourly to multi-day lags) and system levels (flight, airport, route, and neighbor networks), making them highly expressive but also potentially correlated. As a result, they provide strong predictive power but require careful regularization or tree-based modeling to manage redundancy.

Initial exploratory analysis shows that operational features often exhibit the strongest direct relationship with departure delay probability, particularly those representing congestion, lagged delay rates, and aircraft tail effects.

### Feature Family 3: Meteorological Features

The meteorological feature set was engineered to capture both instantaneous weather conditions and cumulative atmospheric effects over multiple time horizons, recognizing that flight delays are often driven not only by current conditions but also by recently evolving weather patterns. This category integrates base station weather measurements, severe weather event indicators, neighboring station signals, fallback-imputed observations, and rolling-window aggregations to provide a comprehensive view of environmental risk.

At the core level, the features include primary weather variables such as precipitation, wind speed, gust intensity, temperature, humidity, visibility, sky cover, and ceiling height. These variables directly capture local atmospheric conditions that affect airport and flight operations. In addition, binary severe-weather flags such as rain, snow, fog, thunderstorm, freezing conditions, haze, and ceiling presence help isolate discrete weather events that are known to cause sharp spikes in operational disruption.

To improve robustness against missing station observations, the feature engineering pipeline incorporates neighbor-based weather features and fallback weather signals, allowing the model to leverage nearby station measurements when direct observations are unavailable. This significantly strengthens resilience in sparse or incomplete weather records.

A key enhancement in this section is the creation of rolling weather aggregation features across 3-hour, 6-hour, 12-hour, and daily windows. These time-series features summarize cumulative precipitation, maximum wind gust, minimum visibility, severe event persistence, and other rolling statistics. Such aggregations are designed to capture the lagged and compounding impact of deteriorating weather, which often influences delays more strongly than a single instantaneous observation.

To summarize the large number of weather-derived variables in an interpretable way, we further introduce composite weather severity indices at each time scale:

* weather_index_3h
* weather_index_6h
* weather_index_12h
* weather_index_daily

These indices consolidate the most operationally relevant weather signals into a single severity score, making it easier to assess how cumulative meteorological conditions influence delay probability.

### Feature Family 4: Graph-Based Features

The graph feature set is designed to capture the structural role of airports within the national flight network, extending beyond local operational or weather conditions to model system-wide connectivity effects. Unlike temporal or meteorological features, graph-based features describe how an airport is positioned within the broader aviation network and how that position influences its exposure to congestion and delay propagation.

At the core of this feature group are network connectivity measures, including inbound and outbound degree, which quantify how many direct flight connections an airport maintains. These features serve as a proxy for airport traffic intensity and operational load, where highly connected airports are expected to experience higher congestion pressure.

In addition, centrality-based measures such as PageRank, hub score, authority score, and betweenness centrality are used to capture different dimensions of influence within the network. PageRank and hub/authority scores identify airports that are structurally important within the flow of traffic, while betweenness centrality highlights airports that act as critical intermediaries or bottlenecks in multi-leg routing paths. These metrics are particularly useful for identifying systemic points of delay amplification.

The feature set also includes weighted neighbor delay propagation metrics, which combine network structure with historical delay behavior of connected airports. These features help quantify how delays propagate through the network rather than remaining isolated to individual nodes, capturing spillover effects that are not observable from local features alone.

Overall, graph features provide a system-level perspective on delay dynamics, shifting the analysis from isolated airport behavior to network-driven interactions. This allows the model to capture whether delays are influenced not only by local conditions, but also by an airport’s position in a highly connected, central, or bottleneck-prone part of the aviation network.

### Feature Family 5: Time-Series Features (RFM)

The time-series feature set is designed to capture the temporal dynamics of flight delays, modeling how operational conditions evolve over time and how historical system states influence current delay risk. Unlike static or structural features, these variables explicitly represent temporal dependence, short-term congestion memory, and evolving system stability, which are critical for understanding delay propagation in aviation networks.

At the core of this feature group are short-term operational delay signals, including delay intensity within 1-hour, 3-hour, and 6-hour windows, as well as related congestion indicators such as flight volume and maximum observed delay in recent periods. These features capture the immediate operational state of an airport, which is shown to be the strongest predictor of near-term delay risk, particularly within the 3-hour window.

The feature set also includes lag-based delay history variables, spanning 1-day to 30-day windows. These features model the memory effect of the system, where past delays continue to influence current conditions but with diminishing strength over time. This decay pattern reflects the gradual normalization of operations after disruption events.

In addition, trend and volatility features provide a higher-level view of system behavior. Trend features capture directional shifts in delay rates over time, while volatility measures quantify instability in operational performance. Together, they help distinguish between consistently congested airports and those experiencing fluctuating disruption patterns.

Finally, operational sequencing and congestion context features, such as departure order and scheduled flight density, represent the structural temporal load on the system. These variables capture how scheduling pressure and queue effects contribute to nonlinear increases in delay probability under high-demand conditions.

Overall, time-series features provide a multi-resolution view of delay dynamics, spanning immediate operational conditions, medium-term memory effects, and long-term behavioral trends. This enables the model to capture not just whether delays occur, but how they evolve and propagate through time under varying levels of system stress.

### Final Feature Selection: From 132 Engineered Features to 20 Model-Ready Predictors

The custom join pipeline initially produced **237 raw and engineered columns** across the five feature families (temporal, operational, meteorological, graph-based, and time-series). Converting this broad feature space into a compact, high-signal predictor set for modeling required a systematic three-stage reduction process.

#### Stage 1: Column Pruning (237 → 132 features)

The first reduction removed columns that are either non-predictive by construction or operationally unavailable at prediction time:

* **Identifier and metadata columns** (`flight_id`, `OP_CARRIER_AIRLINE_ID`, `origin_lat`, `dest_lon`, `origin_station_id`, `dest_station_id`, `latest_weather_ts`, `sched_dep_ts`, `cutoff_ts`, etc.) — these carry no generalizable signal and would cause data leakage or overfitting if retained.
* **Post-departure / target-leaking columns** (`DEP_DELAY`, `CRS_ELAPSED_TIME`, `CANCELLED`, `CANCELLATION_CODE`, `DIVERTED`) — these are only known after the event we aim to predict.
* **High-null neighbor and fallback columns** (19 neighbor-station weather features with >9% missing, plus 16 coalesced fallback columns) — imputing these to zero would bias regularization coefficients. Dropping them reduced noise without losing primary-station weather signal.

After this stage, **\~132 numeric features** remained, spanning all five feature families with representation from direct weather observations, multi-window weather accumulations, operational delay propagation, graph centrality metrics, and time-series RFM signals.

#### Stage 2: Regularization-Based Importance Ranking (132 → ranked list)

To identify which of the 132 features carry the strongest predictive signal for `DEP_DEL15`, we applied **dual-regularization feature selection** using PySpark MLlib's `LogisticRegression` on a stratified 2015 training sample:

1. **L1 (Lasso) Regularization** (`elasticNetParam=1.0`, `regParam=0.01`): L1 drives uninformative coefficients to exactly zero, performing implicit feature selection. Of the 55 numeric features evaluated, only **5 retained non-zero coefficients** — `tail_avg_delay_10h`, `route_avg_delay_6h`, `departure_sequence`, `num_flights_3h`, and `lag_7d_delay_rate` — confirming these as the strongest individual predictors.

2. **L2 (Ridge) Regularization** (`elasticNetParam=0.0`, `regParam=0.01`): L2 shrinks all coefficients toward zero without eliminating them, enabling a continuous ranking of feature importance by absolute coefficient magnitude. This complementary view surfaces features that contribute moderate but collectively meaningful signal — such as `HourlyRelativeHumidity`, `avg_morning_delay_rate_airport`, and `graph_betweenness` — which L1 zeroed out individually but which improve ensemble model performance when combined.

The combined L1+L2 ranking was produced by merging both coefficient vectors and sorting by absolute magnitude, ensuring the final selection captures both **strong individual predictors** (L1-surviving) and **synergistic contributors** (L2-ranked).

#### Stage 3: Top-20 Selection (ranked list → 20 features)

The top 20 features were selected from the combined ranking, with additional considerations:

* **Feature family diversity**: The final set includes at least one representative from each of the five families (temporal, operational, meteorological, graph-based, time-series), ensuring the model captures complementary delay mechanisms rather than over-indexing on a single signal type.
* **Redundancy removal**: Highly correlated feature pairs (e.g., `dep_delay_1h` and `dep_delay_3h`) were deduplicated, retaining only the variant with the stronger regularization coefficient.
* **Computational tractability**: Limiting to 20 features ensures that all four model architectures (LR, XGBoost, MLP, LSTM) can train within the cluster's memory and time budget across the full 5-year sliding-window CV framework.

#### Bridge to Modeling Pipeline

The 20 selected features were assembled into a single Spark ML `features` vector column using `VectorAssembler` and persisted as the pre-split train/test parquet files at `dbfs:/student-groups/Group_3_1/custom-join/all/data_separated/`. All downstream models (Logistic Regression, XGBoost, MLP, LSTM) consume this identical 20-feature vector, ensuring a fair apples-to-apples comparison across architectures. The feature vector construction and train/test split are performed once upstream; no additional feature engineering occurs inside the modeling notebooks.

This separation of concerns — broad feature generation in the EDA/engineering phase, systematic reduction via regularization, and a frozen feature vector for modeling — ensures reproducibility, prevents data leakage, and allows each model to be evaluated solely on its architectural merits.

## Leakage Analysis

### What Is Leakage?

Data leakage occurs when information from outside the training dataset is used to create the model, resulting in overly optimistic performance estimates that do not generalize to real-world deployment. Leakage can take many forms:

* **Target leakage:** Using features that are derived from or correlated with the target variable in ways that would not be available at prediction time (e.g., using actual departure delay to predict whether a flight will be delayed).
* **Temporal leakage:** Training on data from the future to predict the past, violating the causal ordering of events.
* **Train-test contamination:** Allowing information from the test set to influence model training through shared preprocessing, feature computation, or hyperparameter selection.

#### Hypothetical Example of Leakage

Suppose we compute a feature `airport_avg_delay_rate` as the average delay rate at each origin airport across the *entire* dataset (including 2019 test data). A model trained on 2015–2018 data would implicitly "see" information from 2019 through this global statistic, inflating its apparent performance. The correct approach is to compute such features only from historical data strictly prior to each flight's departure time.

### Leakage Audit of Our Pipeline

We conducted a systematic audit of every stage in our pipeline to verify the absence of leakage:

#### 1. Temporal Split Integrity
* **Train:** 2015–2018 (with sliding window cross-validation within this range)
* **Validation:** Chronological holdout within 2015–2018 (October of each year)
* **Test:** 2019 (blind, never consulted during training or tuning)
* **Verification:** The 2019 test set was loaded separately and never joined, sampled, or aggregated with training data at any pipeline stage.

#### 2. Feature Computation Boundaries
* **Weather features:** Aligned using a 2-hour pre-departure cutoff. Weather observations are only included if their timestamp is *before* the scheduled departure time minus 2 hours. No post-departure or concurrent weather data is used.
* **Delay propagation features** (e.g., `cum_delay_rate_today`, `tail_avg_delay_10h`, `route_avg_delay_6h`): Computed using only flights that departed *before* the current flight's scheduled departure. We do NOT use DEP_DELAY, ARR_DELAY, or any actual delay column as input features—only the binary DEP_DEL15 from prior flights.
* **Graph features** (PageRank): Computed on the airport connectivity graph derived from 2015–2018 flight data only. The 2019 test set graph structure is not used.
* **Time-series features** (RFM): Rolling windows look backward only, using strict `< current_timestamp` filtering.

#### 3. Preprocessing and Imputation
* All imputation statistics (means, medians, modes) are computed **exclusively from training data** and applied to validation/test sets.
* Scaling parameters (StandardScaler) are fit on training data only.
* One-hot encoding vocabularies are derived from training data only; unseen categories in test data are mapped to "other."

#### 4. Undersampling Isolation
* Fixed random undersampling is applied within both training and validation partitions, but not on test partitions.
* We use a fixed random undersampling strategy because we want a 50/50 split of the majority/minority class from the initial 80/20 split, to ensure that the model does not fit purely on the majority class. However, this sampling should have been applied to the training data only. Our current sampling happens on the validation data as well, and this results in lower precision for all models when we compare validation and test results. In future work, undersampling should be performed within each sliding-window CV fold, applied only to the training partition while preserving the natural class distribution in the validation partition.

#### 5. Target Variable Isolation
* DEP_DEL15 (the target) is never used as an input feature.
* Related leaky columns (DEP_DELAY, DEP_DELAY_NEW, ARR_DELAY, ARR_DELAY_NEW, CARRIER_DELAY, WEATHER_DELAY, NAS_DELAY, SECURITY_DELAY, LATE_AIRCRAFT_DELAY) are all excluded from the input feature set.
* No arrival-time or actual departure-time features are used, as these would constitute future information.

#### 6. Graph Variables
* One limitation in the current set of graph features is that they use airport-level weighted edge features derived using global deduplicated airport summaries across the full training horizon. 
* Although rolling windows were backward-looking, the use of the latest airport record across multiple years may introduce temporal leakage by allowing earlier observations to inherit future-informed graph states. 
* This may artificially strengthen predictive performance, particularly for graph-based propagation signals. 
* In future work, these features should be reconstructed dynamically at each prediction timestamp using strictly time-bounded rolling airport states.

### Cardinal Sins of ML: Compliance Check

| Cardinal Sin | Our Pipeline Status |
| --- | --- |
| Training on test data | Strict temporal split; 2019 never seen during training |
| Using the target as a feature | DEP_DEL15 and all delay-amount columns excluded from features |
| Future data in features | All features use only pre-departure historical data with 2-hour buffer |
| Data snooping in preprocessing | Imputation/scaling stats computed on training data only |
| Leaky cross-validation | Sliding window temporal CV; no random shuffling across time |
| Evaluation on training data | All reported metrics are from held-out validation or blind test |

## Modeling Pipelines

### Methodology and Design

Phase III expands the modeling framework from a single Logistic Regression baseline to three distinct modeling pipelines, each designed to capture different aspects of the delay prediction problem:

1. **Logistic Regression (L1 Regularized):** Continued baseline from Phase II, now trained on the custom join dataset with expanded features.
1. **Logistic Regression (L2 Regularized):** Continued baseline from Phase II, now trained on the custom join dataset with expanded features.
2. **XGBoost:** Ensemble of decision trees for capturing nonlinear feature interactions and providing built-in feature importance rankings.
3. **Multilayer Perceptron (MLP) Neural Network:** Deep learning approach for learning complex decision boundaries that linear or tree-based models may miss.

All three pipelines share the same evaluation framework established in Phase II:
* **Temporal split:** Train on 2015–2018, test on 2019 (blind)
* **Sliding window time-series cross-validation** within 2015–2018
* **Fixed random undersampling** on the full training set prior to folding (1:1 balance); XGBoost uses class weighting (`scale_pos_weight`) on the full dataset instead
* **F2-optimized threshold tuning** on validation set
* **Evaluation metrics:** F2 Score (primary), Precision, Recall

### Modeling Pipeline Visualization

**Figure Description:** The Phase III modeling pipeline is a multi-stage architecture designed to ensure temporal integrity, class balance, and rigorous model comparison. The pipeline proceeds as follows:

1. **Data Ingestion and Custom Join:** In addition to using pre-joint OTPW 60 months dataset, raw OTPW flight records are joined with station-level weather observations, producing a unified dataset spanning 2015–2019 with approximately 78 engineered features across five families (Temporal, Meteorological, Operational, Graph-Based, and Time-Series/RFM).

2. **Temporal Train-Test Split:** The dataset is partitioned chronologically—2015–2018 for training/validation and 2019 as a fully blind test set. This strict temporal boundary prevents any future information from leaking into model training.

3. **Sliding Window Time-Series Cross-Validation:** Within the 2015–2018 training period, five sequential folds are constructed using expanding training windows and fixed-length validation windows (one quarter each). Each fold respects temporal ordering, ensuring that validation data always follows training data chronologically.

4. **Fixed Random Undersampling:** On the full 2015–2018 training set prior to cross-validation folding, the majority class (on-time flights, \~80%) is randomly downsampled to achieve approximate 1:1 parity with the minority class (delayed flights, \~20%). This prevents the majority class from dominating during gradient optimization. Because undersampling is applied before folding, both training and validation partitions within each CV fold reflect the rebalanced distribution (\~52% delayed) rather than the natural \~20% rate. Test partitions are preserved at their natural distributions to ensure unbiased final evaluation. XGBoost follows a different strategy: it trains on the full non-undersampled dataset and addresses class imbalance through the `scale_pos_weight` parameter, which reweights the loss function to penalize missed delays more heavily without discarding majority-class samples. We acknowledge that 1:1 downsampling can remove some meaningful information from the majority class. We believe this can be improved by changing our downsampling strategy from strict 1:1 downsampling to more advanced techniques that better preserve the integrity of the data while still reducing its volume. For example, stratified sampling could help us preserve the structure of the original data by maintaining proportional representation across important groups, reservoir sampling could allow us to keep a representative random sample efficiently when working with very large data like ours, and systematic sampling could reduce the data in a simpler way by selecting every nth observation. Another possible solution would be to apply variable downsampling rates. In that case, we could adjust the downsampling rate dynamically based on metric behavior and business importance. For example, higher sampling rates could be used during peak periods or anomalous conditions, while lower rates could be used during more stable periods.

5. **Parallel Model Training:** Three model families are trained independently on each cross-validation fold:
   - *Logistic Regression (L2 Regularized):* Linear baseline with Ridge penalty for coefficient shrinkage.
   - *XGBoost:* Gradient-boosted ensemble for capturing nonlinear feature interactions and providing built-in feature importance rankings.
   - *Multilayer Perceptron (MLP):* Neural network with sigmoid activations for learning complex, non-separable decision boundaries.

6. **F2-Optimized Threshold Tuning:** For each model, the classification threshold is swept from 0.10 to 0.90 on the validation set. The threshold maximizing the F2 score (which weights recall twice as heavily as precision) is selected, reflecting the business priority of minimizing missed delay predictions.

7. **Blind Test Evaluation (2019):** The best-performing configuration from each model family is applied to the held-out 2019 dataset. Final metrics (F2, Precision, Recall) are reported on this unseen data to provide an unbiased estimate of generalization performance.

This architecture ensures that no information flows backward in time, all preprocessing decisions are fit exclusively on training data, and model comparison is conducted under identical experimental conditions.

### Loss Functions

#### Logistic Regression
The Logistic Regression pipeline minimizes **Binary Cross-Entropy (Log Loss)** with **L2 Regularization** (Ridge):

$$J(w) = - \sum_{i=1}^{N} \left[ y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right] + \frac{\lambda}{2} \|w\|_2^2$$

#### XgBoost
XGBoost minimizes a regularized objective function. For binary classification, it typically uses **Binary Cross-Entropy (Log Loss)** as the prediction loss, while split selection at each tree node is based on gain, which measures the reduction in the objective after a split:

$$mathcal{L}^{(t)} = \sum_{i=1}^{N} l\left(y_i, \hat{y}_i^{(t-1)} + f_t(x_i)\right) + \Omega(f_t)$$

$$Omega(f_t) = \gamma T + \frac{1}{2}\lambda \sum_{j=1}^{T} w_j^2$$

$$text{Gain} = \frac{1}{2}\left(\frac{G_L^2}{H_L+\lambda} + \frac{G_R^2}{H_R+\lambda} - \frac{(G_L+G_R)^2}{H_L+H_R+\lambda}\right) - \gamma$$

#### Multilayer Perceptron (MLP)
The MLP minimizes **Binary Cross-Entropy** with backpropagation:

$$L = - \frac{1}{N} \sum_{i=1}^{N} \left[ y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right]$$

With sigmoid activation in the output layer and ReLU activations in hidden layers. The Spark MLP Classifier uses L-BFGS or gradient descent for optimization.

#### LSTM (Long Short-Term Memory)
The LSTM model minimizes **Binary Cross-Entropy with Logits Loss** (`BCEWithLogitsLoss`), which combines a sigmoid activation and binary cross-entropy into a single numerically stable operation:

$$L = - \frac{1}{N} \sum_{i=1}^{N} \left[ y_i \log(\sigma(z_i)) + (1 - y_i) \log(1 - \sigma(z_i)) \right]$$

where \\(z_i\\) is the raw logit output from the final fully connected layer and \\(\sigma(z_i) = \frac{1}{1 + e^{-z_i}}\\) is the sigmoid function applied internally by the loss.

The model is optimized using **Adam** (Adaptive Moment Estimation), which maintains per-parameter adaptive learning rates via first and second moment estimates of the gradients. Training leverages **AMP (Automatic Mixed Precision)** on GPU with `GradScaler` for numerically stable FP16 forward passes, and employs **early stopping** (patience=5 epochs) monitored on validation AUC-PR to prevent overfitting.

### Tackling Class Imbalance: Fixed Random Undersampling

The class imbalance strategy remains consistent with Phase II. Flight departure delays exhibit a natural skew of approximately 80% on-time to 20% delayed. We apply fixed Random Undersampling on the full 2015–2018 training set prior to cross-validation folding, forcing a 1:1 balance to prevent majority-class bias. Because undersampling is applied before the temporal CV split, both training and validation partitions within each fold are drawn from the already-rebalanced pool. Test partitions are preserved at their natural distributions to ensure unbiased evaluation.

It is important to note that the undersampling strategy differs by model family. For Logistic Regression, MLP, and LSTM, the majority class is downsampled to approximate 1:1 parity on the full training set before folding. As a result, validation metrics reflect balanced-class conditions (~52% delayed) rather than the natural ~80/20 distribution. This partially explains the precision gap observed between CV validation and the 2019 blind test set, where the natural 20% delay rate produces substantially more false positives at the same decision threshold. XGBoost, by contrast, trains on the full non-undersampled dataset and addresses class imbalance through the `scale_pos_weight` parameter, which reweights the loss function to penalize missed delays more heavily without discarding majority-class samples.

### Evaluation Metrics

Consistent with Phase II, the **F2 Score** serves as our primary evaluation metric. The F2 score weights recall twice as heavily as precision, reflecting the business priority of identifying delayed flights for proactive operations management.

$$Precision = \frac{TP}{TP + FP}$$

$$Recall = \frac{TP}{TP + FN}$$

$$F_2 = \frac{5 \cdot Precision \cdot Recall}{4 \cdot Precision + Recall}$$

Where:
* \\(TP\\) = True Positives (correctly predicted delays)
* \\(FP\\) = False Positives (false alarms)
* \\(FN\\) = False Negatives (missed delays)
* \\(P_k\\), \\(R_k\\) = Precision and Recall at the \\(k-th\\) decision threshold

### Results Discussion

**Key Findings:**

1. **Both datasets converge on L1 regularization** (`elasticNetParam=1.0`) with minimal penalty (`regParam=0.001`), suggesting that feature sparsity (eliminating uninformative features) is more beneficial than coefficient shrinkage (L2) for this prediction task.

2. **OOF vs. Test performance gap:** Both models show high OOF metrics (\\(F_2\\) > 0.80) but lower test \\(F_2\\) (~0.75). This degradation is primarily driven by **precision collapse** on the 2019 test set (Precision drops from ~0.87 in OOF to ~0.40 on test), while recall remains high (>0.95). This asymmetry indicates a temporal distribution shift: the 2019 class distribution differs from 2015–2018 training data, causing the model to over-predict delays.

3. **Threshold sensitivity to temporal shift:** The Custom Join model's OOF-optimized threshold (0.25) further amplifies the precision-recall trade-off on 2019 data. The OTPW model's decision to retain the default threshold (0.5) reflects a pragmatic trade-off: accepting slightly lower recall in exchange for better precision stability across temporal boundaries.

4. **OTPW outperforms Custom Join on test \\(F_2\\)** (0.7615 vs. 0.7481) despite lower OOF metrics, suggesting the MVP feature set generalizes more robustly to unseen temporal periods. The Custom Join's richer feature set (20 vs. 18 features) provides better in-sample discrimination (higher AUC-PR) but may overfit to historical temporal patterns.

5. **Computational efficiency:** The OTPW pipeline is ~8× faster end-to-end (11.4 min vs. 90.8 min), driven by faster HP tuning (8.3 vs. 31.8 min) — likely due to sparser feature interactions in the simpler feature set and larger data volume that parallelizes more efficiently.

## Neural Network (MLP)

As requested by the client, we implemented a **Multilayer Perceptron (MLP) Neural Network** using PySpark's `MultilayerPerceptronClassifier`. We experimented with three different network architectures to evaluate the impact of network depth on prediction performance.

We tested three specific architectures because they provide enough representational power to capture nonlinear relationships and feature interactions that a linear model cannot capture, while still being compact enough to reduce the risk of overfitting and keep training computationally manageable. The decreasing structure is a standard practice, since it encourages the network to learn progressively more compact representations before the final binary output layer. In contrast, a one-hidden-layer network may be too limited to model the complexity of interactions among feature buckets, while much larger architectures would add parameters quickly, increasing training time and overfitting risk without a clear reason to expect better performance on this dataset size and problem type.

We are using this architectures to understand whether an increase in capacity improves performance. 
We start from 16, and then move to 32 and 64 nodes that are common baseline widths in neural networks. The second hidden layers with 8, 16 and 32 nodes respectively preserve a simple funnel structure, which is a common heuristic for classification tasks. Since this is not an image or language problem, there is usually no need to start with very deep or very wide architectures. 

#### Architecture 1: 

```
Input Layer (20 features) → Hidden Layer (16 neurons, 8 neurons, sigmoid) → Output Layer (2 classes, softmax)
layers = [20, 16, 8, 2]
```


#### Architecture 2:

```
Input Layer (20 features) → Hidden Layer (32 neurons, 16 neurons, sigmoid) → Output Layer (2 classes, softmax)
layers = [20, 32, 16, 2]
```

#### Architecture 3:

```
Input Layer (20 features) → Hidden Layer (64 neurons, 32 neurons, sigmoid) → Output Layer (2 classes, softmax)
layers = [20, 64, 32, 2]
```

### Results Discussion

1. **Comparing Architectures**: Across both datasets, increasing model complexity (more neurons per layer) does not translate into meaningful performance gains. On the MVP dataset, CV F2 improves only marginally from 0.8930 (MLP-1) to 0.8978 (MLP-3), but test F2 actually decreases from 0.7164 to 0.7091. On the custom dataset, MLP-1 already achieves the best test F2 (0.8256), while complex models (MLP-2 and MLP-3) slightly underperform. The simplest architecture (MLP-1) generalizes better. Additional complexity adds noise-fitting capacity without improving real predictive power.

2. **Overfitting Behavior**: There are clear signs of overfitting as complexity increases. CV performance slightly improves with deeper networks, but test performance remains the same or even declines. For example, in the MVP dataset:
MLP-3 has the highest CV F2 (0.8978) but the lowest test F2 (0.7091). Similarly, in the custom dataset: MLP-3 achieves the highest CV F2 (0.9040) but not the best test result. This pattern suggests that the model is learning patterns specific to training folds, not generalizable structure.
Increasing depth leads to variance increase without bias reduction.

3. **Computational Trade-off**: The cost of complexity is clear as the MLP-3 is 2.5 to 4x slower than MLP-1. There is no justification for using deeper architectures given the sharp increase in training time and no improvement in outcomes.

4. **MVP vs Custom Dataset (Feature Engineering Impact)**: The custom dataset reduces false positives significantly, improving precision. The model becomes more balanced, instead of over-predicting delays.

5. **Trade-offs of Moving to Custom Dataset**: Transitioning to the custom-joined dataset delivers a clear improvement in predictive performance, particularly in precision and F2-score, driven by a richer feature set. However, these gains come with important considerations for production deployment. The use of a larger dataset (5 years vs. 1 year) increases storage and processing requirements, while the expanded feature space (20 vs. 12 variables) adds complexity to the model and results in moderately longer training times. Although these computational costs remain manageable for simpler architectures.


## LSTM (Long Short-Term Memory)

As an additional deep learning approach beyond the required MLP, we implemented an **LSTM neural network** using PyTorch, trained on a Tesla T4 GPU with Automatic Mixed Precision (AMP). The LSTM was selected to explore whether recurrent architectures designed to capture sequential dependencies could improve delay prediction when applied to tabular flight features. Both dataset configurations use the same temporal CV framework as the other models: 5 sliding window folds (6Q train / 2Q validation, 2Q step), grid search over hyperparameter combinations, OOF prediction aggregation, \\(F_2\\)-optimized threshold tuning, and final blind test on 2019.

---

### Architecture

The `FlightDelayLSTM` model consists of:
* **Input:** 20 features (Custom Join) or 18 features (OTPW), reshaped to a single-timestep sequence `[batch, 1, features]`
* **LSTM layers:** 2 stacked LSTM layers with `hidden_size=128` and inter-layer `dropout=0.3`
* **Output head:** Dropout(0.3) → Linear(128, 1) → raw logit (sigmoid applied by `BCEWithLogitsLoss`)
* **Trainable parameters:** 209,025 (Custom Join, 20 features) / 208,001 (OTPW, 18 features)

## Ensemble Model (Stacking)

Flight delay prediction is inherently a nonlinear, multi-factor, and temporally dependent problem, influenced by operational congestion, weather dynamics, carrier behavior, and network-level interactions. While individual models such as Logistic Regression (LR), XGBoost (XGB), MLP, and LSTM capture different aspects of this complexity, no single model consistently performs best across all scenarios.

To improve predictive robustness, an ensemble learning approach using stacking was adopted. The motivation behind stacking is to combine the strengths of multiple heterogeneous models:

* Logistic Regression (LR): Strong baseline, interpretable, captures linear signal structure
* XGBoost (XGB): Captures nonlinear interactions and threshold effects effectively
* MLP / LSTM (planned but partially excluded): Designed to capture deep nonlinear and temporal dependencies

The goal of the ensemble is to optimize recall-oriented performance (F2 score) while maintaining generalization across diverse operational conditions.

Due to computational and time constraints, fold-based stacking was implemented only for LR and XGBoost, as full fold-model training pipelines were available for these models. MLP and LSTM were trained as final models without fold-level OOF generation, and therefore were not included in the stacking layer.

### Stacking Strategy

A two-level stacking architecture was implemented to enhance flight delay prediction by leveraging complementary strengths of multiple models while improving generalization and recall performance.

**1. Level 1 (Base Models)**

This includes independently trained models; Logistic Regression, XGBoost, and Long Short-Term Memory (LSTM) that serve as primary learners over a rich feature space comprising temporal patterns, operational congestion signals, weather-driven indices, and graph-based network features. 

  Each model captures complementary aspects of the delay-generation process: LR provides an interpretable and stable linear baseline, XGBoost captures complex nonlinear interactions and decision thresholds, while LSTM complements both by learning temporal persistence, sequential dependencies, and dynamic delay propagation across time windows.

**2. At Level 2 (Meta Model)**
 
 This includes a Logistic Regression meta-learner is trained on out-of-fold (OOF) predictions generated from the Level 1 models. These OOF predictions ensure unbiased learning by reflecting model performance on unseen temporal segments. The meta-model uses these probabilistic outputs as input features to learn optimal combinations of base model signals, effectively learning when to trust each model under varying operational conditions.

This stacking formulation enables the system to dynamically weight model contributions, correct systematic biases across base learners, and improve sensitivity to rare but operationally critical flight delay events, thereby optimizing overall predictive robustness.

### Fold Strategy

A time-aware sliding window cross-validation strategy was used for model training and stacking, ensuring strict temporal consistency across all stages of the pipeline. Importantly, the exact same folding strategy used during baseline model training (LR and XGBoost) was reused during OOF generation for stacking. This alignment was intentional to ensure that the fold models produced during base model training could be directly and correctly leveraged without retraining or mismatch in data partitions.

Each fold consists of a training window of 6 quarters (~1.5 years) followed by a validation window of 2 quarters, with a step size of 2 quarters between successive folds. This rolling structure preserves chronological ordering and simulates real-world forecasting conditions, where only past information is available at prediction time.

By maintaining identical fold definitions across both base model training and stacking-level OOF prediction generation, the pipeline ensures full consistency between learned model parameters and their corresponding validation outputs. This design prevents temporal leakage, enables reliable reuse of fold-trained models, and guarantees that stacking inputs are constructed from truly out-of-fold predictions. Consequently, it strengthens the validity of the meta-learning stage while maintaining realistic temporal generalization across the entire system.

### Ensemble Pipeline

The proposed ensemble system follows a two-stage stacking architecture designed to improve flight delay prediction performance by combining complementary strengths of multiple machine learning models.

![](https://raw.githubusercontent.com/Pinkaban/FlightDelayPrediction/refs/heads/main/Final%20Report%20Notebooks/ReportImages/Phase%20III%20Report%20Images/w261_project_phase3_sthakur_ensemble_pipeline_updated.png)

The pipeline begins with a shared training dataset (4 years) and a holdout test dataset (1 year). Three primary base models; Logistic Regression (LR), XGBoost (XGB), and Long Short-Term Memory (LSTM) are used to build this pipeline. These models are trained using a time-aware sliding window strategy to generate robust and leakage-free learning.

In the first stage, fold-based models (OOF predictions) are used to construct a stacking dataset. These out-of-fold predictions from LR, XGB, and LSTM capture model behavior on unseen temporal segments, ensuring unbiased training signals for the meta-learner. In parallel, final trained models are retained for test-time inference.

In the second stage, a meta-feature layer is constructed using the predicted probabilities from the base models. A Logistic Regression meta-model is then trained on these features to learn optimal combinations of base model outputs and improve overall predictive stability.

During inference, the same base models generate probability outputs on the test dataset, which are combined into a unified stacking feature representation. The meta-model processes these inputs to produce the final ensemble probability and prediction.

Finally, the model outputs are evaluated using the holdout test set to assess generalization performance and operational effectiveness in predicting flight delays.

### Threshold Optimization

Threshold selection plays a critical role in translating probabilistic model outputs into operationally meaningful decisions. While the ensemble model generates calibrated probabilities for flight delay risk, the final classification performance is highly sensitive to the probability cutoff used to assign the positive class (delay).

Given the problem objective of **minimizing missed delay events**, the threshold optimization process was guided primarily by the **F-beta score with β = 2 (F2 score)**, which places greater emphasis on recall than precision. The initial threshold sweep indicated that the **maximum F2 score occurred at a low threshold of approximately 0.10**, resulting in near-perfect recall. However, this setting also substantially increased false positives, causing the model to overpredict delays and potentially generate excessive operational alerts.

![](https://raw.githubusercontent.com/Pinkaban/FlightDelayPrediction/refs/heads/main/Final%20Report%20Notebooks/ReportImages/Phase%20III%20Report%20Images/w261_project_phase3_sthakur_ensemble_threshold_optimization.png)

Although this threshold is mathematically optimal under the F2 objective, it is not necessarily the most practical choice from a business standpoint. In airline operations, while **false negatives (missed delays)** carry a high cost and must be minimized, **false positives (incorrect delay alerts)** also impose operational burden through unnecessary interventions, scheduling noise, and alert fatigue. Excessive false alarms can reduce trust in the model and limit its usability in real-world decision support systems.

A more balanced threshold analysis showed that a **threshold near 0.50** still maintained strong recall performance (approximately 90%+) while significantly improving precision and reducing false alerts. This provides a more operationally robust trade-off between sensitivity and alert quality. Therefore, rather than selecting the threshold solely based on the absolute maximum F2 score, the final decision boundary was chosen based on a **business-aligned trade-off: maximizing precision subject to maintaining a high recall target**.

This thresholding strategy ensures that the ensemble model remains highly sensitive to delay events while preserving practical usability and decision-making reliability. It also reflects a cost-aware modeling philosophy, where statistical optimization is aligned with real-world operational impact rather than metric maximization alone.

### Discussion of Results

The comparative evaluation of the four modeling approaches—Logistic Regression (LR), XGBoost (XGB), LSTM, and the final stacking-based Ensemble model—provides strong evidence that the ensemble framework delivers the most balanced, robust, and operationally valuable performance.

At the baseline level, Logistic Regression demonstrates exceptionally high recall (0.9642), indicating strong sensitivity to delay events and a very low false-negative rate. This makes LR highly effective at identifying potential delays and minimizing missed disruption events. However, this performance comes at the cost of significantly lower precision (0.4136), meaning that a substantial proportion of flights flagged as delayed are false positives. While this makes LR valuable as a high-sensitivity screening model, its relatively low precision can lead to excessive false alerts and reduced operational trust.

In contrast, XGBoost substantially improves the balance between precision and recall. With a precision of 0.6310, recall of 0.8960, and F2 score of 0.8260, it demonstrates strong nonlinear learning capability and improved control of false positives while still maintaining high recall. This indicates that XGBoost effectively captures complex interactions across operational, weather, graph-based, and temporal engineered features that are not fully represented through linear decision boundaries.

The LSTM model introduces an additional and highly complementary perspective by explicitly modeling sequential temporal dependencies and delay propagation behavior over time. Its recall reaches an exceptionally high 0.9961, the highest among all models, demonstrating near-perfect sensitivity to delay events. This strongly validates the importance of temporal persistence and lag-driven dynamics in flight delay prediction. However, similar to LR, this high recall is accompanied by relatively lower precision (0.4601), indicating a tendency toward over-prediction and increased false-positive alerts. Nevertheless, its strong F2 score (0.8079) confirms that temporal sequence learning contributes significant predictive value.

The final stacking ensemble model, which integrates LR, XGBoost, and LSTM through a Logistic Regression meta-learner, delivers the strongest overall performance. With a precision of 0.6253, recall of 0.9057, F1 score of 0.7398, and the highest F2 score of 0.8311, the ensemble achieves the best recall-weighted performance across all evaluated models.

The key value brought by the ensemble lies in its ability to combine the complementary strengths of all three base learners:

- LR contributes strong linear baseline sensitivity
- XGBoost contributes nonlinear interaction learning and sharper decision boundaries
- LSTM contributes sequential temporal memory and delay propagation awareness

This allows the meta-learner to dynamically learn when to rely on:

- the strong sensitivity of LR,
- the structured nonlinear decision capability of XGB, and
- the temporal persistence signals captured by LSTM.

As a result, the ensemble retains recall levels close to the strongest individual models while substantially improving precision and overall balance.

From a business perspective, this is highly significant. The final ensemble provides strong delay detection capability without overwhelming operations with excessive false alarms, making it substantially more practical for deployment in real-world decision-support systems such as:

- proactive flight disruption management
- airport resource planning
- gate and crew scheduling optimization
- customer communication workflows
- delay escalation monitoring

Most importantly, the improvement is not merely incremental from a statistical perspective. The ensemble meaningfully improves the trade-off between sensitivity, reliability, and operational usability, making it the most deployment-ready model among all evaluated approaches.

### Phase II vs Phase III: Logistic Regression

The migration from the OTPW dataset to the custom join dataset represents the most significant data-level improvement in this project. The custom join provides:

* Multi-window weather accumulations capturing evolving conditions
* Network delay propagation features from the airport connectivity graph

The Phase II best Logistic Regression achieved \\(F_2\\) = 0.5500 on the OTPW dataset with 18 features. In Phase III, the same Logistic Regression architecture on the OTPW 5-year dataset achieved \\(F_2\\) = 0.7615 on the 2019 blind test (a **+0.2115 improvement**) attributable to the expanded training window (5 years vs. 1 year) and the temporal CV framework that better captures seasonal patterns. The Custom Join LR achieved \\(F_2\\) = 0.7481, slightly lower than OTPW LR, suggesting that for a linear model the simpler MVP feature set generalizes more robustly to unseen temporal periods, even though the Custom Join's richer features provide better in-sample discrimination (OOF \\(F_2\\) 0.8328 vs. 0.8030).

#### Impact of Graph and Time-Series Features

The Custom Join dataset incorporates graph-based features (airline PageRank, airport degree centrality, route betweenness, hub score) and time-series RFM features (delay recency, frequency, monetary averages) that are absent from the OTPW dataset. The impact of these features is most visible in the nonlinear models:

* **LSTM on Custom Join** (test \\(F_2\\) = 0.8332 at default threshold) **vs. LSTM on OTPW** (test \\(F_2\\) = 0.7245): a **+0.1087 gain**, demonstrating that the graph and time-series features provide substantial additional signal when a model can capture nonlinear interactions.
* **XGBoost on Custom Join** (test \\(F_2\\) = 0.826) **vs. XGBoost on OTPW** (test \\(F_2\\) = 0.732): a **+0.094 gain**, confirming that tree-based ensembles effectively exploit the heterogeneous feature families through nonlinear splits.
* **MLP on Custom Join** (test \\(F_2\\) = 0.8256) **vs. MLP on OTPW** (test \\(F_2\\) = 0.7164): a **+0.109 difference**, demonstrates that hat the enhanced feature set provides strong nonlinear separability even for relatively simple neural architectures, improving the model’s ability to capture complex interactions without requiring additional depth.
* **LR on Custom Join** (test \\(F_2\\) = 0.7481) **vs. LR on OTPW** (test \\(F_2\\) = 0.7615): a **−0.0134 difference**, indicating that the linear model cannot fully exploit the richer feature set and actually slightly overfits to temporal patterns in training.


This divergence confirms that the value of graph and time-series features is **model-dependent**: nonlinear architectures (LSTM, XGBoost, MLP) can leverage the complex feature interactions, while linear models see diminishing or negative returns from the added complexity.

## Conclusion

This project set out to address the high-impact problem of predicting flight departure delays of 15 minutes or more, progressing from exploratory analysis in Phase I to a baseline modeling framework in Phase II, and ultimately to a significantly more advanced data and modeling pipeline in Phase III.

In Phase II, we fitted a Logistic Regression trained on the 1-year OTPW dataset, achieving an F2 score of 0.5500. While this result demonstrated the feasibility of delay prediction using engineered features, it was constrained by both the limited feature set and the linear nature of the model. The OTPW dataset, though useful, relied on pre-joined and relatively coarse weather information, limiting the model’s ability to capture complex temporal and operational dynamics.

By expanding the dataset to five years and engineering a rich set of features—including temporal patterns, weather dynamics, network propagation effects, and time-series delay behavior—we achieved a substantial improvement in predictive performance. The best individual model, XGBoost, reached an F2 score of 0.826 on the 2019 blind test set, representing a +0.276 improvement over Phase II. This gain highlights the combined impact of better data integration, richer feature representation, and more expressive models.

After evaluating multiple architectures (Logistic Regression, XGBoost, MLP and LSTM), we selected a stacked ensemble model combining Logistic Regression and XGBoost as the final solution. This model achieved the highest overall performance, with an F2 score of 0.8285, precision of 0.6185, and recall of 0.9054. The selection reflects a balance between predictive power and robustness: XGBoost contributes strong nonlinear learning and feature interaction capture, while Logistic Regression adds stability and complementary signal through its linear structure. This combination proved effective in improving generalization on unseen data.

Overall, the results demonstrate that combining high-quality, domain-informed feature engineering with appropriate modeling techniques can significantly improve delay prediction accuracy. ## Gap Analysis of Best Pipeline

This section presents a systematic gap analysis of our best-performing pipeline: the **Stacking Ensemble (LR + XGBoost + LSTM → LR Meta-Learner)** trained on the **Custom Join dataset** with 20 engineered features, evaluated on the 2019 blind test set at threshold = 0.50.

### Current State Summary

| Dimension | Achieved Value |
| :--- | :--- |
| Primary Metric (Test \\(F_2\\)) | **0.8311** |
| Test Precision | 0.6253 |
| Test Recall | 0.9057 |
| Test F1 | 0.7398 |
| AUC-ROC | 0.9603 |
| CV \\(F_2\\) (OOF, threshold=0.5) | ~0.907 (base model avg) |
| Total Pipeline Wall Time | ~600+ min (LR + XGB + LSTM + Ensemble) |
| Feature Count | 20 (selected from 237 engineered) |



#### Gap 1: Precision–Recall Imbalance

| Metric | CV (OOF) | 2019 Test | Gap |
| :--- | :---: | :---: | :---: |
| Precision | ~0.87 (XGB base) | 0.6253 | **−0.245** |
| Recall | ~0.90 (XGB base) | 0.9057 | −0.005 |
| \\(F_2\\) | ~0.907 | 0.8311 | −0.076 |

**What we observe:** Precision collapses by ~25 percentage points from cross-validation to the 2019 blind test, while recall remains stable. The ensemble generates approximately 545K–1.2M false positives on the test set (depending on base model), meaning roughly 1 in 3 delay predictions is a false alarm.

**Root cause:** Temporal distribution shift between the 2015–2018 training period and 2019. The delay rate, carrier mix, route network, and weather patterns in 2019 differ from the training distribution. All models — from linear LR to deep LSTM — exhibit this same precision collapse, confirming it is a **data distribution problem, not a model architecture problem**.

**Why it matters:** In operational deployment (gate scheduling, crew planning, passenger communication), a 37.5% false alarm rate erodes trust in the prediction system and wastes operational resources on flights that depart on time.

**Closing the gap:**
* Implement **online calibration** or **sliding-window recalibration** using the most recent 30–90 days of observed outcomes to adapt the decision boundary to current delay rates
* Explore **domain adaptation techniques** (e.g., importance weighting, adversarial training) to make the model robust to temporal covariate shift
* Deploy with **confidence-tiered alerts** (high/medium/low risk) rather than binary predictions, allowing operations teams to filter based on prediction certainty



#### Gap 2: Temporal Generalization and Concept Drift

**What we observe:** The OOF-optimized threshold of 0.20 (which maximizes \\(F_2\\) on cross-validation) consistently **underperforms** the default threshold of 0.50 on the 2019 test set across all models. For the LSTM, OOF threshold=0.20 yields test \\(F_2\\) = 0.8079 vs. 0.8332 at threshold=0.50 (a −0.025 gap). This counterintuitive result means that threshold optimization on historical data can actually **hurt** future performance.

**Root cause:** The temporal CV folds (2015–2018) share more distributional similarity with each other than with 2019. Threshold tuning overfits to the training-era class balance (~20% delay rate), but the 2019 delay rate and feature distributions have shifted. The sliding window CV strategy, while sound, cannot fully anticipate future distributional changes.

**Why it matters:** Any production system that periodically re-tunes its threshold on historical data may inadvertently degrade performance when the operating environment changes.

**Closing the gap:**
* Implement **monitoring for data drift** (feature distribution tests, prediction distribution tracking) with automated alerts when drift exceeds a threshold
* Use **adaptive thresholding** that adjusts based on a rolling window of recent predictions and observed outcomes, rather than fixed historical optimization
* Evaluate **conformal prediction** methods that provide distribution-free coverage guarantees regardless of shift magnitude


#### Gap 3: Feature Coverage Limitations

**What we observe:** The top 20 features were selected from 237 engineered variables. While they capture temporal, operational, weather, graph, and time-series signals, several potentially high-value feature families were excluded:

| Missing Feature Category | Potential Impact | Reason for Exclusion |
| :--- | :--- | :--- |
| Destination weather | High — arrival weather affects turnaround delays | Not available in current join pipeline |
| Carrier-level trend features (7d/30d) | Medium-High — carrier reliability is the #1 delay cause | Excluded during feature reduction |
| Holiday proximity score | Medium — +28% delay rate in peak periods | Not retained in final 20 |
| Real-time ATC/NOTAM data | High — airspace restrictions cause cascading delays | Not available in dataset |
| Aircraft-specific features (age, type, utilization) | Medium — mechanical delays are carrier-driven | Not available in dataset |

**Root cause:** The feature reduction pipeline (237 → 132 → 20) optimized for a compact, low-redundancy set. Some features with moderate individual signal were dropped despite potentially contributing complementary information in nonlinear models. Additionally, several high-value data sources (destination weather, ATC data, aircraft metadata) were not available in the source datasets.

**Why it matters:** The XGBoost feature importance analysis shows that `DISTANCE` alone accounts for 40,267 gain (3× the next feature). This heavy reliance on a single feature suggests the model may benefit from richer, more diverse signals.

**Closing the gap:**
* Reintegrate **carrier_delay_rate_7d** and **holiday_proximity** features, which showed strong EDA signal but were not retained
* Build a **destination weather join** pipeline mirroring the origin weather approach
* Explore **feature interaction terms** (e.g., carrier × time-of-day, weather severity × hub status) that XGBoost may not fully capture through individual splits


#### Gap 4: Computational Cost vs. Marginal Gain

| Model | Test \\(F_2\\) | Wall Time | \\(F_2\\) per Hour |
| :--- | :---: | :---: | :---: |
| MLP-1 | 0.8256 | 6 min | 8.256 |
| XGBoost | 0.8260 | ~360 min | 0.138 |
| LSTM | 0.8332 | 186 min | 0.269 |
| Ensemble | 0.8311 | ~600+ min | 0.083 |

**What we observe:** The ensemble requires training all three base models plus the meta-learner (~600+ minutes total pipeline), yet it improves over the best individual model (LSTM at threshold=0.5, \\(F_2\\) = 0.8332) by only **−0.002 \\(F_2\\) points** — it does not even exceed LSTM. Meanwhile, MLP-1 achieves 99.3% of the ensemble's performance in 1% of the time.

**Root cause:** The base models share significant redundancy in their correct predictions (all capture the same "easy" delay cases). The meta-learner has limited capacity to improve beyond what the strongest base model already achieves. The stacking gains are concentrated in the precision-recall trade-off (ensemble precision = 0.625 is better-balanced than LSTM's 0.46 at threshold=0.20) rather than in raw \\(F_2\\) improvement.

**Why it matters:** For production retraining cadences (daily/weekly), a 600+ minute pipeline creates infrastructure and cost constraints. If the ensemble does not meaningfully outperform XGBoost or MLP, the complexity may not be justified.

**Closing the gap:**
* Consider a **simpler two-model ensemble** (LR + XGBoost only) which avoids GPU-dependent LSTM training while retaining complementary signal
* Investigate **model distillation**: train the full ensemble once, then distill its predictions into a single lightweight model (e.g., XGBoost or MLP) that approximates ensemble behavior at a fraction of the cost
* Benchmark **MLP-1 as the production candidate** given its 60× speed advantage and negligible \\(F_2\\) gap (0.8256 vs. 0.8311)


#### Gap 5: Class Imbalance Strategy

**What we observe:** The pipeline uses fixed random undersampling to address the ~80/20 class imbalance (on-time vs. delayed). While this is a reasonable baseline approach, it:
* Discards a substantial portion of the majority class, potentially losing informative negative examples
* Uses a fixed sampling ratio rather than adapting to fold-specific class distributions
* Does not leverage cost-sensitive learning, which could achieve similar recall without discarding data

**Root cause:** Undersampling was chosen for simplicity and consistency with Phase II. More sophisticated imbalance strategies (SMOTE, class weights, focal loss) were not systematically evaluated.

**Why it matters:** The discarded majority-class samples may contain boundary cases that are critical for improving precision — exactly the metric where the pipeline is weakest.

**Closing the gap:**
* Evaluate **class-weighted loss functions** (e.g., `scale_pos_weight` in XGBoost, class weights in LR/MLP) that preserve all training data while up-weighting the minority class
* Test **SMOTE or ADASYN** oversampling of the minority class as an alternative to majority undersampling
* Implement **threshold-moving** as a post-hoc technique rather than resampling-based approaches

#### Gap 6: Single Holdout Year Evaluation

**What we observe:** All final test metrics are based on a single holdout year (2019). While the temporal CV provides robust validation within 2015–2018, the test evaluation is a single-point estimate with no confidence interval.

**Root cause:** The dataset spans only 2015–2019. Using 2019 as the sole test year is the most rigorous temporal evaluation possible given the data, but it means all reported test metrics are contingent on 2019-specific conditions (e.g., weather patterns, carrier operations, route changes unique to that year).

**Why it matters:** If 2019 was an unusually easy or hard year for delay prediction, the reported \\(F_2\\) = 0.8311 may not generalize to other years. There is no way to estimate the variance of test performance across years from a single holdout.

**Closing the gap:**
* Extend the dataset to **2020–2024** and perform rolling-year evaluation (train on years 1–N, test on year N+1) to estimate performance variance across years
* Report **bootstrap confidence intervals** on the 2019 test set to quantify uncertainty in the reported metrics
* Conduct **stratified analysis** by quarter, carrier, and airport to identify whether the model performs consistently or has pockets of weakness

#### Final Insights

The best pipeline achieves strong delay detection (\\(F_2\\) = 0.8311, recall = 0.9057) and excellent ranking performance (AUC-ROC = 0.9603), representing a **+0.281 \\(F_2\\) improvement** over the Phase II baseline. However, the primary gap, precision collapse under temporal distribution shift remains the most critical barrier to production deployment. Closing this gap requires moving beyond static model training toward **adaptive, drift-aware prediction systems** that continuously recalibrate to current operating conditions. The secondary gaps (feature coverage, compute efficiency, imbalance strategy) represent incremental improvement opportunities that could collectively push \\(F_2\\) beyond 0.85 while reducing false alarm rates to operationally acceptable levels.
